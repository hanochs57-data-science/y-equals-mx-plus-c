```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r'housing - housing.csv')
print(df.head())
```

```python
print(df.select_dtypes(include="number").columns)
```

```python
correlation = df.corr(numeric_only=True)
print(correlation)
```

```python
plt.Figure(figsize=(10,7))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import LabelEncoder,StandardScaler

df = pd.read_csv(r'house_price_regression_dataset - house_price_regression_dataset.csv')
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

Label_encoder = LabelEncoder()

for col in [col for col in df.columns if df[col].dtype=='object']:
    df[col] = Label_encoder.fit_transform(df[col])

X = df.drop(columns=['House_Price'])
Y = df[['House_Price']]

scaler = StandardScaler()
X = scaler.fit_transform(X)

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, Y_train)
Y_pred = model.predict(X_test)

print("RMSE: ",root_mean_squared_error(Y_pred,Y_test))
print("MSE: ",mean_absolute_error(Y_pred,Y_test))
print("R^2 Score: ",r2_score(Y_pred,Y_test))

plt.Figure(figsize=(10,7))
sns.regplot(data = df,x = "Square_Footage",y = "House_Price")
plt.xlabel("Square_Footage")
plt.ylabel("House_Price")
plt.show()
print("HANOCH SHETTY T043")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import LabelEncoder,StandardScaler

df = pd.read_csv(r'housing - housing.csv')
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

Label_encoder = LabelEncoder()

for col in [col for col in df.columns if df[col].dtype=='object']:
    df[col] = Label_encoder.fit_transform(df[col])

X = df[['housing_median_age','total_rooms','population','households','median_income']]
Y = df[['median_house_value']]

scaler = StandardScaler()
X = scaler.fit_transform(X)

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, Y_train)
Y_pred = model.predict(X_test)

print("RMSE: ",root_mean_squared_error(Y_pred,Y_test))
print("MSE: ",mean_absolute_error(Y_pred,Y_test))
print("R^2 Score: ",r2_score(Y_pred,Y_test))

plt.Figure(figsize=(10,7))
sns.regplot(data = df,x = "housing_median_age",y = "median_house_value")
plt.xlabel("housing_median_age")
plt.ylabel("median_house_value")
plt.show()
print("HANOCH SHETTY T043")
```
